```python
data = np.load(path_2)
```

```python
print("Data shape:", data.shape)
print("Data type:", data.dtype)
print("Data sample:", data[:10])  # Print the first 10 samples to check
```

```python
import numpy as np
import matplotlib.pyplot as plt

points_per_waveform = 16

# Number of complete waveforms
num_waveforms = len(data) // points_per_waveform

print("Number of waveforms:", num_waveforms)

# Remove any leftover points if length is not exactly divisible by 16
data_trimmed = data[:num_waveforms * points_per_waveform]

# Reshape into shape: (num_waveforms, 16)
waveforms = data_trimmed.reshape(num_waveforms, points_per_waveform)

# Max amplitude of each waveform
```

```python
from scipy.signal import resample

wf = waveforms[:1_000_000].astype(np.int32)
waveforms_squared = wf ** 2
max_amplitudes = np.max(wf, axis=1)
max_indices = np.argmax(wf, axis=1)


gradients = np.gradient(wf, axis=1)

# Allocate slew-rate array

slew_rate_first_to_max = np.full(wf.shape[0], np.nan)

# Loop over waveforms

for i in range(wf.shape[0]):

    idx_max = max_indices[i]

    # skip waveforms where max is already at first point

    if idx_max > 0:

        # gradient only from first point to max point

        slew_rate_first_to_max[i] = np.max(gradients[i, :idx_max + 1])

print("Slew rate shape:", slew_rate_first_to_max.shape)

print("First 10 slew rates:", slew_rate_first_to_max[:10])



area = np.trapz(wf, axis=1)
area_squared = np.trapz(waveforms_squared, axis=1)

N = 1000000

interp_factor = 10

n_original = wf.shape[1]          # 16

n_interp = n_original * interp_factor  # 160

# Allocate result

max_amp_sinc = np.empty(N)

# Process in batches to avoid huge memory usage

batch_size = 50_000

for start in range(0, N, batch_size):

    end = min(start + batch_size, N)

    # sinc/Fourier interpolation along each waveform

    wf_interp = resample(wf[start:end], n_interp, axis=1)

    # max amplitude of interpolated waveform

    max_amp_sinc[start:end] = np.max(wf_interp, axis=1)



print("Max amplitudes shape:", max_amplitudes.shape)
print("First 10 max amplitudes:", max_amplitudes[:10])
print("Area shape:", area.shape)
print("First 10 areas:", area[:10])
```

```python
plt.figure(dpi=300)
plt.hist(max_amplitudes, bins=800)
plt.xlabel("Maximum amplitude")
plt.ylabel("Counts")
plt.title("Maximum Amplitude Discrimination")
plt.xscale('log')
plt.show()
```

```python
plt.figure(dpi=300)
plt.hist(area, bins=800)
plt.xlabel("Area of Squared Waveforms")
plt.ylabel("Counts")
plt.title("Area Of Squared Waveforms Discrimination")
plt.xscale('log')
plt.show()
```

```python
plt.figure(dpi=300)
plt.hist(slew_rate_first_to_max, bins=800)
plt.xlabel("Slew Rate")
plt.ylabel("Counts")
plt.title("Slew Rate Discrimination")
plt.xscale('log')
plt.show()
```

```python
plt.figure(dpi=300)
plt.hist(max_amp_sinc, bins=800)
plt.xlabel("Maximum amplitude (sinc interpolated)")
plt.ylabel("Counts")
plt.title("Maximum Amplitude Discrimination (sinc interpolated)")
plt.xscale('log')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def positive_only(x):
    """
    Keep only finite and positive values for log-x histograms.
    """
    x = np.asarray(x)
    return x[np.isfinite(x) & (x > 0)]


# Prepare data for log-x histograms
# max_amp_pos = positive_only(max_amplitudes)
# area_pos = positive_only(area)
# area_squared_pos = positive_only(area_squared)
# slew_pos = positive_only(slew_rate_first_to_max)
# max_amp_sinc_pos = positive_only(max_amp_sinc)

# Figure with 6 panels in one frame
fig, axes = plt.subplots(2, 3, figsize=(18, 10), dpi=300)

axes = axes.ravel()

bins = 700

# 1. Max amplitude
axes[0].hist(max_amplitudes, bins=bins)
axes[0].set_xscale("log")
axes[0].set_xlabel("Max amplitude")
axes[0].set_ylabel("Counts")
axes[0].set_title("Max Amplitude Discrimination")
axes[0].grid(True)

# 2. Area
axes[1].hist(area, bins=bins)
axes[1].set_xscale("log")
axes[1].set_xlabel("Area")
axes[1].set_ylabel("Counts")
axes[1].set_title("Area Discrimination")
axes[1].grid(True)

# 3. Area squared
axes[2].hist(area_squared, bins=bins)
axes[2].set_xscale("log")
axes[2].set_xlabel("Area of squared waveform")
axes[2].set_ylabel("Counts")
axes[2].set_title("Squared Area Discrimination")
axes[2].grid(True)

# 4. Slew rate first-to-max
axes[3].hist(slew_rate_first_to_max, bins=bins)
axes[3].set_xscale("log")
axes[3].set_xlabel("Slew rate first-to-max")
axes[3].set_ylabel("Counts")
axes[3].set_title("Slew Rate Discrimination")
axes[3].grid(True)

# 5. Sinc interpolated max amplitude
axes[4].hist(max_amp_sinc, bins=bins)
axes[4].set_xscale("log")
axes[4].set_xlabel("Sinc-interpolated max amplitude")
axes[4].set_ylabel("Counts")
axes[4].set_title("Sinc Max Amplitude Discrimination")
axes[4].grid(True)

# 6. Superposed max amplitude vs sinc-interpolated max amplitude
axes[5].hist(
    max_amplitudes,
    bins=bins,
    label="Original max amplitude"
)

axes[5].hist(
    max_amp_sinc,
    bins=bins,
    label="Sinc-interpolated max amplitude"
)

axes[5].set_xscale("log")
axes[5].set_xlabel("Amplitude")
axes[5].set_ylabel("Counts")
axes[5].set_title("Max Amplitude vs Sinc Max Amplitude Discrimination")
axes[5].legend()
axes[5].grid(True)

fig.suptitle("Photon-Number Discrimination Histograms", fontsize=16)

plt.tight_layout()
plt.savefig("photon_number_discrimination_histograms.png", dpi=300)
plt.show()
```
